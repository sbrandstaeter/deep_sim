```python
h = (0.5, 0.8)
n = [6, 7, 8, 9, 10]
space = Space([h, n])
n_samples = 10

x = space.rvs(n_samples)
```

```python
def plot_searchspace(x, title, h, n):
    fig, ax = plt.subplots()
    plt.plot(np.array(x)[:, 0], np.array(x)[:, 1], 'bo', label='samples')
    plt.plot(np.array(x)[:, 0], np.array(x)[:, 1], 'bo', markersize=1, alpha=0.5)
    # ax.legend(loc="best", numpoints=1)
    ax.set_xlabel("X1")
    ax.set_xlim([h[0], h[-1]])
    ax.set_ylabel("X2")
    ax.set_ylim([n[0], n[-1]])
    plt.title(title)
```

```python
plot_searchspace(x,"some",h,n)
```

```python
lhs = Lhs(criterion="maximin", iterations=10000)
x = lhs.generate(space.dimensions, n_samples)
```

```python
plot_searchspace(x,"some",h,n)
```

```python
x_array = np.array(x)

#for h_v in h:
#    print(f"h value: {h_v}, the counts: {sum(x_array[:,0]==h_v)}")
print("\n")
for n_v in n:
    print(f"n value: {n_v}, the counts: {sum(x_array[:,1]==n_v)}")
```

```python
hammersly = Hammersly()
x = hammersly.generate(space.dimensions, n_samples)

x_array = np.array(x)

for h_v in h:
    print(f"h value: {h_v}, the counts: {sum(x_array[:,0]==h_v)}")
print("\n")
for n_v in n:
    print(f"n value: {n_v}, the counts: {sum(x_array[:,1]==n_v)}")
```
