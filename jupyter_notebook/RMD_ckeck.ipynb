```python
import numpy as np
from  numpy import random as rnd    
from scipy.stats import norm
```

```python
def generate_2D_surface(n, H, zref):
        '''
        creates the 2D surfaces using RMD (Random Midpoint Distribution)

        '''
        np.random.seed(1)

        N = 2**n     
        z = np.zeros([N+1,N+1])

        alpha = 1 / np.sqrt(0.09)

        D = N
        d = N//2

        for _ in range(n):
            alpha=alpha/np.sqrt(2)**H
            
            for j in range(d,N-d+1,D):
                for k in range(d,N-d+1,D):
                    z[j,k] =  (z[j+d,k+d]+z[j+d,k-d]+z[j-d,k+d]+z[j-d,k-d])/4+alpha*norm.ppf(np.random.rand())
            
            alpha=alpha/np.sqrt(2)**H
            
            for j in range(d,N-d+1,D):
                z[j,0]=(z[j+d,0]+z[j-d,0]+z[j,d])/3+alpha*norm.ppf(np.random.rand())
                z[j,N]=(z[j+d,N]+z[j-d,N]+z[j,N-d])/3+alpha*norm.ppf(np.random.rand())
                z[0,j]=(z[0,j+d]+z[0,j-d]+z[d,j])/3+alpha*norm.ppf(np.random.rand())
                z[N,j]=(z[N,j+d]+z[N,j-d]+z[N-d,j])/3+alpha*norm.ppf(np.random.rand())
    
            for j in range(d,N-d+1,D):
                for k in range(D,N-d+1,D):
                    z[j,k]=(z[j,k+d]+z[j,k-d]+z[j+d,k]+z[j-d,k])/4+alpha*norm.ppf(np.random.rand())
            
            for j in range(D,N-d+1,D):
                for k in range(d,N-d+1,D):
                    z[j,k]=(z[j,k+d]+z[j,k-d]+z[j+d,k]+z[j-d,k])/4+alpha*norm.ppf(np.random.rand())


            D = D//2
            d = d//2 

        scalefactor = zref/(np.max(z)-np.mean(z))
        z = z*scalefactor
        z = z-(np.min(z))
        return z

```

```python
z = generate_2D_surface(2, 0.1, 50)

print(z.mean())
print(z.std())
print(z.max())
```
