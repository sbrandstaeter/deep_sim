```python
n = 9
path = "/tmp/pytest-of-a11btasa/pytest-44/test_rs_iterator0/"

df = pd.read_csv(path+"contact_area.dat", header=None)

for i in range(1):
    name = path + "surface_" + str(i) + ".dat"

    z_surf = np.loadtxt(fname=name,delimiter=";",usecols=range(2**n+1))
    print(f"surface: {name}")
    print(f"contact area: {df.iloc[i,1]}")
    print(f"mean:\t {z_surf.mean()}")
    print(f"std:\t {z_surf.std()}")
    print("\n")
```
